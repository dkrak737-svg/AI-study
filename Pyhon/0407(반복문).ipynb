### `for` 문과 함께 리스트 슬라이싱 사용하기 예시

````python
my_list = [10, 20, 30, 40, 50]

# 리스트의 첫 3개 요소에 대해 반복
print("리스트의 첫 3개 요소:")
for item in my_list[0:3]:
    print(item)

# 리스트의 마지막 2개 요소에 대해 반복
print("\n리스트의 마지막 2개 요소:")
for item in my_list[-2:]:
    print(item)

# 짝수 인덱스의 요소에 대해 반복 (0, 2, 4...)
print("\n짝수 인덱스 요소:")
for item in my_list[::2]:
    print(item)

# 홀수 인덱스의 요소에 대해 반복 (1, 3, 5...)
print("\n홀수 인덱스 요소:")
for item in my_list[1::2]:
    print(item)
````

158. 파일 이름이 저장된 리스트에서 확장자를 제거하고 파일 이름만 화면에 출력하라. (힌트: split() 메서드)

````python
리스트 = ['hello.py', 'ex01.py', 'intro.hwp']

for i in 리스트:
  split = i.split('.') # . 기준으로 분리해서 변수에 담고
  print(split[0]) # 그 변수의 [0]번째 출력
````

159. 파일 이름이 저장된 리스트에서 확장자가 .h인 파일 이름을 출력하라.


````python
리스트 = ['intra.h', 'intra.c', 'define.h', 'run.py']

for i in 리스트:
  split = i.split('.')
  if split[-1] == 'h':
    print(split[0])
````
