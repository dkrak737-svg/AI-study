```python
# 숫자를 하나 바아서 uswer에 저장
user = input("")

# 대 소문자 전환
if user.islower():    # 만약 user 값이 소문자라면
  print(user.upper()) # 대문자로 전환
else:                 # 그렇지 않다면
  print(user.lower()) # 소문자로 전환
```

```python
score = int(input())  # 숫자 하나 입력받아 score에 저장

if 81 <= score <= 100: # 81 ~100이면 A
  print("A")
if 61 <= score <= 80:  # 61 ~ 80이면 B
  print("B")
if 41 <= score <= 60:  # 41 ~ 60이면 C
  print("C")
if 21 <= score <= 40:  # 21 ~ 40이면 D
  print("D")
if 0 <= score <= 20:   # 0 ~ 20이면 E
  print("E")

```

```python
# 화폐 종류와, 금액 입력받기
money = input("화폐 종류를 임력하시오 : ")
cost = float(input("금액을 입력하시오 : "))
# user = input("입력: ")
# num, currency = user.split() >> 이런 방법도 있음


# 화폐 종류와 숫자가 적힌 사전 만들기
exchange = {"달러" : 1167,
            "엔" : 1.096,
            "유로" : 1268,
            "위안" : 171}

# 조건문 사용하여 환율 계산해 출력하기
if money == "달러":
  print(exchange["달러"] * cost)
elif money == "엔":
  print(exchange["엔"] * cost)
elif money == "유로":
  print(exchange["유로"] * cost)
elif money == "위안":
  print(exchange["위안"] * cost)

# print(float(num) * 환율[currency], "원")
# >> 이런 방법도 있음
```

```python
# 전화번호 입력받고, '-' 기준으로 스플릿해 분리하기
number = input("전화번호를 입력 하시오(xxx-xxxx-xxxx) : ")
num = number.split("-")[0]

# 조건문으로 [0]번째 문자열일 경우 통신사를 com에 저장
if num == '011':
  com = "SKT"
elif num == '016':
  com = "KT"
elif num == '019':
  com = "LG"
else:
  com = "알 수 없음"

# 출력
print(f"당신은 {com} 사용자입니다.")
```

```python
# 주민등록번호 입력받고 남자 여자 판별
# 입력받고, '-'기준으로 스플릿
number = input()
num = number.split('-')[1]

# 조건문으로 앞자리 숫자별 성별 출력
if num[0] == 1 or 3:
  print('남자')
elif num[0] == 2 or 4:
  print('여자')
```

```python
# 라이브러리 가져오기
import requests

# .get 사용하여 btc 사전데이터 가져오기
btc = requests.get("https://api.bithumb.com/public/ticker/").json()['data']

# btc 사전의 밸류값 변수명에 저장(자료형은 float)
변동폭 = float(btc['max_price']) - float(btc['min_price'])
시가 = float(btc['opening_price'])
최고가 = float(btc['max_price'])

# 조건문 달아서 상승장 하락장 판별하여 출력
if (시가+변동폭) > 최고가:
  print("상승장")
else:
  print("하락장")


```
