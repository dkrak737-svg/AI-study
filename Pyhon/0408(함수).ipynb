234. 숫자로 구성된 하나의 리스트를 입력받아, 짝수들을 추출하여 리스트로 반환하는 pickup_even 함수를 구현하라.  
```
pickup_even([3, 4, 5, 6, 7, 8])
```
```
[4, 6, 8]
```

````python
def pickup_even(list):
   new_list = []
   for i in list:
      if i % 2 == 0:
        new_list.append(i)
   print(new_list)


pickup_even([3, 4, 5, 6, 7, 8])
````

```

```

````python
def convert_int(str):
  split_str = str.split(',')
  # split_str은 ['1', '234', '567']과 같은 리스트입니다.
  # 리스트의 요소들을 빈 문자열로 연결하여 '1234567'을 만듭니다.
  # 그리고 이것을 정수로 변환합니다.
  print(int(''.join(split_str)))

  #'기준문자'.join(리스트)
  #''아무것도 넣지 않으면 아무것도 사이에 끼지 않고 합침


convert_int("1,234,567")
````

220. 세 개의 숫자를 입력받아 가장 큰수를 출력하는 print_max 함수를 정의하라. 단 if 문을 사용해서 수를 비교하라.

````python
def print_max(x, y, z):
  if x > y and x > z:
    print(x)
  elif y > z and y > x:
    print(y)
  else:
    print(z)

print_max(7, 15, 120)

def max_val(a, b, c):
  max_val = 0
  if a > max_val:
    max_val = a
  if b > max_val:
    max_val = b
  if c > max_val:
    max_val = c
  print(max_val)

max_val(7, 150, 120)
````
