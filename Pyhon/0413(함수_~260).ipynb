인스턴스 변수에 접근하여 값을 가져오는 예

258. 클래스 메소드 - 2
사람 (Human) 클래스에 (이름, 나이, 성별)을 받는 setInfo 메소드를 추가하세요.
```
>>> areum = Human("모름", 0, "모름")
>>> areum.setInfo("아름", 25, "여자")
```

````python
class Human:
  def __init__(self, name, age, sex):
    self.age = age
    self.name = name
    self.sex = sex

  def who(self):
    print(f"이름: {self.name}, 나이: {self.age}, 성별: {self.sex}")

  def setinfo(self, name, age, sex): # 업데이트, #번수에 새로 받아주
    self.age = age
    self.name = name
    self.sex = sex

areum = Human("모름", 0, "모름")
areum.who()
````

259.
사람 (human) 클래스에 "나의 죽음을 알리지 말라"를 출력하는 소멸자를 추가하세요.
```
areum = Human("아름", 25, "여자")
del areum
나의 죽음을 알리지 말라
```

````python
class Human:
  def __init__(self, name, age, sex):
    self.age = age
    self.name = name
    self.sex = sex

  def who(self):
    print(f"이름: {self.name}, 나이: {self.age}, 성별: {self.sex}")

  def setinfo(self, name, age, sex): # 업데이트, #번수에 새로 받아주
    self.age = age
    self.name = name
    self.sex = sex

  def __del__(self):
    print('나의 죽음을 알리지 마라')

aruem = Human("아름", 25, "여자")
del(aruem)
````
