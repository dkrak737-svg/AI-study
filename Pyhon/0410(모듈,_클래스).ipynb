257. 클래스 메소드 - 1  
사람 (Human) 클래스에서 이름, 나이, 성별을 출력하는 who() 메소드를 추가하세요.
```
>>> areum.who()
이름: 조아름, 나이: 25, 성별: 여자
```

````python
class Human:
    def __init__(self, name, age, sex):
      self.name = name
      self.age = age
      self.sex = sex

    def who(self):
      print(f"이름: {self.name}, 나이: {self.age}, 성별: {self.sex}")

areum = Human('아름', 25, '여자')
areum.who()
````

258. 클래스 메소드 - 2  
사람 (Human) 클래스에 (이름, 나이, 성별)을 받는 setInfo 메소드를 추가하세요.
```
>>> areum = Human("모름", 0, "모름")
>>> areum.setInfo("아름", 25, "여자")
```

````python
class Human:
  def __init__(self, name, age, sex):
    self.name = name
    self.age = age
    self.sex = sex

  def who(self):
    print(f"이름: {self.name}, 나이: {self.age}세, 성별: {self.sex}")

  def setinfo(self, name, age, sex):
    self.name = name
    self.age = age
    self.sex = sex

areum = Human('모름', 25, '모름')
areum.who()

areum.setinfo('아름', 25, '여자')
areum.who()
````

259 클래스 소멸자.  
사람 (human) 클래스에 "나의 죽음을 알리지 말라"를 출력하는 소멸자를 추가하세요.
```
>>> areum = Human("아름", 25, "여자")
>>> del areum
나의 죽음을 알리지 말라
```

````python
class Human:
  def __init__(self, name, age, sex): # 객체 생성시 초기값 설정(어던 값을 받고 저장할지)
    self.name = name
    self.age = age
    self.sex = sex

  def who(self):
    print(f"이름: {self.name}, 나이: {self.age}세, 성별: {self.sex}")

  def setinfo(self, name, age, sex): # 값 변경 >> 덮어씀
    self.name = name
    self.age = age
    self.sex = sex

  def __del__(self): # 소멸자 >> 객체 삭제할 때 사용
    print("나의 죽음을 알리지 말라")

areum = Human("아름", 25, "여자")
del areum
````
